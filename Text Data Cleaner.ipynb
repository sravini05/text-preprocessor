```python
from collections import Counter
import string
import re

# Regex to remove emojis
emoji_pattern = re.compile("["
                           u"\U0001F600-\U0001F64F"  # emoticons
                           u"\U0001F300-\U0001F5FF"  # symbols & pictographs
                           u"\U0001F680-\U0001F6FF"  # transport & map symbols
                           u"\U0001F1E0-\U0001F1FF"  # flags
                           u"\U00002700-\U000027BF"  # dingbats
                           u"\U000024C2-\U0001F251"
                           "]+", flags=re.UNICODE)

while True:
    print("\n===== TEXT CLEANER MENU =====")
    print("1) STOPWORDS REMOVE")
    print("2) STEMMING")
    print("3) LEMMATIZATION")
    print("4) REMOVE PUNCTUATION")
    print("5) REMOVE DUPLICATE WORDS")
    print("6) CLEAN EXTRA SPACES")
    print("7) SAVE TEXT TO FILE")
    print("8) WORD FREQUENCY COUNT")
    print("9) TOP 3 FREQUENT WORDS")
    print("10) REMOVE EMOJIS")
    print("0) EXIT")

    user_choice = int(input("Enter your choice: "))

    # ---------------- OPTION 1: STOPWORDS ----------------
    if user_choice == 1:
        from nltk.corpus import stopwords
        original_text = input("Enter text: ")
        words_list = original_text.lower().split()
        stop_words_set = set(stopwords.words("english"))
        filtered_words = [word for word in words_list if word not in stop_words_set]

        print("Before:", original_text)
        print("After:", " ".join(filtered_words))

    # ---------------- OPTION 2: STEMMING ----------------
    elif user_choice == 2:
        from nltk.stem import PorterStemmer
        original_text = input("Enter text: ")
        words_list = original_text.lower().split()
        stemmer = PorterStemmer()
        stemmed_words = [stemmer.stem(word) for word in words_list]

        print("Before:", original_text)
        print("After:", " ".join(stemmed_words))

    # ---------------- OPTION 3: LEMMATIZATION ----------------
    elif user_choice == 3:
        from nltk.stem import WordNetLemmatizer
        original_text = input("Enter text: ")
        words_list = original_text.lower().split()
        lemmatizer = WordNetLemmatizer()
        lemmatized_words = [lemmatizer.lemmatize(word) for word in words_list]

        print("Before:", original_text)
        print("After:", " ".join(lemmatized_words))

    # ---------------- OPTION 4: REMOVE PUNCTUATION ----------------
    elif user_choice == 4:
        original_text = input("Enter text: ")
        text_no_punctuation = original_text.translate(str.maketrans('', '', string.punctuation))

        print("Before:", original_text)
        print("After:", text_no_punctuation)

    # ---------------- OPTION 5: REMOVE DUPLICATES ----------------
    elif user_choice == 5:
        original_text = input("Enter text: ")
        unique_words_text = " ".join(dict.fromkeys(original_text.split()))

        print("Before:", original_text)
        print("After:", unique_words_text)

    # ---------------- OPTION 6: CLEAN EXTRA SPACES ----------------
    elif user_choice == 6:
        original_text = input("Enter text: ")
        cleaned_text = " ".join(original_text.split())

        print("Before:", original_text)
        print("After:", cleaned_text)

    # ---------------- OPTION 7: SAVE TEXT TO FILE ----------------
    elif user_choice == 7:
        original_text = input("Enter text: ")
        with open("output.txt", "w", encoding="utf-8") as file:
            file.write(original_text)
        print("Text saved to output.txt")
        print("Saved Text:", original_text)

    # ---------------- OPTION 8: WORD FREQUENCY ----------------
    elif user_choice == 8:
        original_text = input("Enter text: ").lower()
        words_list = original_text.split()
        word_frequency = Counter(words_list)

        print("Text:", original_text)
        print("Word Frequency:", word_frequency)

    # ---------------- OPTION 9: TOP 3 WORDS ----------------
    elif user_choice == 9:
        original_text = input("Enter text: ").lower()
        words_list = original_text.split()
        word_frequency = Counter(words_list)
        top_3_words = word_frequency.most_common(3)

        print("Text:", original_text)
        print("Top 3 Words:", top_3_words)

    # ---------------- OPTION 10: REMOVE EMOJIS ----------------
    elif user_choice == 10:
        original_text = input("Enter text: ")
        text_no_emojis = emoji_pattern.sub(r'', original_text)

        print("Before:", original_text)
        print("After:", text_no_emojis)

    # ---------------- EXIT ----------------
    elif user_choice == 0:
        print("Exiting...")
        break

    # ---------------- INVALID OPTION ----------------
    else:
        print("Invalid option! Please choose a number from 0 to 10.")

    print("-----------------------")
```
